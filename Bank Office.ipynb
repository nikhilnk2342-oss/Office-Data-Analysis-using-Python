```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('ggplot')
%matplotlib inline
```

```python
data = {
    'Employee_ID':[101,102,103,104,105,106,107,108,109,110],
    'Name':['Amit','Riya','Karan','Sneha','Rahul',
            'Priya','Ankit','Neha','Vikas','Pooja'],
    'Department':['HR','IT','Finance','IT','Marketing',
                  'Finance','HR','IT','Marketing','Finance'],
    'Salary':[45000,70000,65000,75000,55000,
              68000,48000,72000,60000,67000],
    'Experience':[2,5,4,6,3,5,2,7,4,6],
    'Attendance':[92,95,88,96,90,93,89,97,91,94],
    'Performance_Score':[75,88,80,92,78,85,74,95,82,89]
}

df = pd.DataFrame(data)
df
```

```python
df.head()
df.info()
df.describe()
```

```python
df.isnull().sum()
```

```python
dept_count = df['Department'].value_counts()
print(dept_count)
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='Department', data=df)
plt.title("Employees in Each Department")
plt.show()
```

```python
avg_salary = df['Salary'].mean()
print("Average Salary:", avg_salary)
```

```python
df[df['Salary']==df['Salary'].max()]
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Salary'], bins=5, kde=True)
plt.title("Salary Distribution")
plt.show()
```

```python
dept_salary = df.groupby('Department')['Salary'].mean()
print(dept_salary)
```

```python
dept_salary.plot(kind='bar')
plt.title("Average Salary by Department")
plt.ylabel("Salary")
plt.show()
```

```python
avg_attendance = df['Attendance'].mean()
print("Average Attendance:", avg_attendance)
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x='Name', y='Attendance', data=df)
plt.xticks(rotation=45)
plt.title("Employee Attendance")
plt.show()
```

```python
df.sort_values(by='Performance_Score', ascending=False).head(1)
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Performance_Score'], kde=True)
plt.title("Performance Score Distribution")
plt.show()
```

```python
corr = df[['Salary','Experience','Attendance',
           'Performance_Score']].corr()

corr
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='Experience', y='Salary',
                hue='Department', data=df)

plt.title("Experience vs Salary")
plt.show()
```

```python
def performance(score):
    if score >= 90:
        return "Excellent"
    elif score >= 80:
        return "Good"
    elif score >= 70:
        return "Average"
    else:
        return "Poor"

df['Performance_Level'] = df['Performance_Score'].apply(performance)

df[['Name','Performance_Level']]
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='Performance_Level', data=df)
plt.title("Performance Categories")
plt.show()
```

```python
print("Total Employees:", len(df))
print("Average Salary:", df['Salary'].mean())
print("Highest Salary:", df['Salary'].max())
print("Average Attendance:", df['Attendance'].mean())
print("Top Performer:",
      df.loc[df['Performance_Score'].idxmax(),'Name'])
```
